```python
import polars as pl
import duckdb
from pathlib import Path

DATA_DIR = Path("../data/raw")
con = duckdb.connect()

def inventory(name, df: pl.DataFrame):
    print(f"\n{'='*60}")
    print(f"FUENTE: {name}")
    print(f"{'='*60}")
    print(f"Filas: {df.height:,} | Columnas: {df.width}")
    print(f"\nSchema:")
    print(df.schema)
    print(f"\nPrimeras filas:")
    print(df.head(3))
    print(f"\nNulos por columna:")
    print(df.null_count())
```

```python
retail = pl.read_csv(DATA_DIR / "retail_orders.csv", infer_schema_length=10000)
inventory("RETAIL", retail)
```

```python
ecomm = pl.read_ndjson(DATA_DIR / "ecomm_orders.ndjson")
inventory("ECOM (dotCom)", ecomm)
```

```python
app = pl.read_json(DATA_DIR / "app_orders.json")
inventory("APP", app)
```

```python
wholesale = pl.read_parquet(DATA_DIR / "wholesale_orders.parquet")
inventory("WHOLESALE", wholesale)
```

```python
franchise = pl.read_excel(DATA_DIR / "franchise_orders.xlsx")
inventory("FRANCHISE", franchise)
```

```python
currency = pl.read_csv(DATA_DIR / "currency_reference.csv")
inventory("CURRENCY REFERENCE", currency)
```

```python
def unique_vals(df: pl.DataFrame, cols: list[str], name: str, n=15):
    print(f"\n{'='*60}\n{name}\n{'='*60}")
    for c in cols:
        if c in df.columns:
            vals = df[c].unique().to_list()
            print(f"\n{c} ({len(vals)} valores únicos):")
            print(vals[:n])

unique_vals(retail, ["store_country", "order_month", "currency", "loyalty_flag", "order_state"], "RETAIL")
unique_vals(ecomm, ["market", "period", "currency", "adiClub", "status"], "ECOMM")
unique_vals(app, ["cc", "yearmonth", "category", "currency", "member", "fulfillment"], "APP")
unique_vals(wholesale, ["country_code", "category_code", "currency", "status"], "WHOLESALE")
unique_vals(franchise, ["Pais", "Mes", "Moneda", "AdiClub", "Estado_Orden"], "FRANCHISE")
```

```python
ecomm_exploded = ecomm.explode("lines").unnest("lines")
print(ecomm_exploded["catCode"].value_counts().sort("catCode"))
print(ecomm_exploded.head())
```

```python
# ¿El article de ecomm se parece a los IDs de producto de otras fuentes?
print("Muestra articles ecomm:", ecomm_exploded["article"].unique().to_list()[:10])
print("Muestra productId app:", app["productId"].unique().to_list()[:10])
print("Muestra material_id wholesale:", wholesale["material_id"].unique().to_list()[:10])

# Si comparten espacio de IDs, probamos el cruce
articles_ecomm = set(ecomm_exploded["article"].unique().to_list())
products_app = set(app["productId"].cast(pl.Utf8).unique().to_list())
materials_wh = set(wholesale["material_id"].unique().to_list())

print("\nOverlap ecomm-app:", len(articles_ecomm & products_app))
print("Overlap ecomm-wholesale:", len(articles_ecomm & materials_wh))
```

```python
# Normalizar categorías de app (corrigiendo typos) antes de cruzar
category_fix = {
    "Runing": "Running", "Running": "Running",
    "Outdor": "Outdoor", "Outdoor": "Outdoor",
    "Footbal": "Football", "Football": "Football",
    "Originls": "Originals", "Originals": "Originals",
}

app_clean = app.with_columns(
    pl.col("category").replace(category_fix).alias("category_clean")
)

# Tabla de referencia: productId -> categoría (de app)
app_ref = app_clean.select(["productId", "category_clean"]).unique()

# Convertir article de ecomm a mismo tipo que productId (Int64) para el join
ecomm_with_cat = ecomm_exploded.with_columns(
    pl.col("article").cast(pl.Int64).alias("productId")
).join(app_ref, on="productId", how="inner")

# Ver la relación catCode <-> category_clean
crosstab = ecomm_with_cat.group_by(["catCode", "category_clean"]).agg(
    pl.len().alias("count")
).sort(["catCode", "count"], descending=[False, True])

print(crosstab)
```
